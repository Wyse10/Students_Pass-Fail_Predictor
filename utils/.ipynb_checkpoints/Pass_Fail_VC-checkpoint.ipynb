```python
def plot_outliers_boxplot(df):

    numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
    num_cols = len(numeric_cols)

    plt.figure(figsize=(15, 5 * (num_cols // 3 + 1)))

    for i, col in enumerate(numeric_cols, 1):
        plt.subplot((num_cols // 3 + 1), 3, i)
        sns.boxplot(x=df[col])
        plt.title(f'Box plot of {col}')

    plt.tight_layout()
    plt.show()
```

```python
plot_outliers_boxplot(df)
```

```python
# Functions to remove outliers in individual columns
def remove_outliers_iqr(df, column, multiplier= 1.5):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - multiplier * IQR
    upper_bound = Q3 + multiplier * IQR

    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
```

```python
df = remove_outliers_iqr(df, "Wassce Grade", 1.5)
```

```python
df = remove_outliers_iqr(df, 'First_Year_CGPA', 1.5)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
#from xgboost import XGBClassifier
from sklearn.metrics import classification_report, accuracy_score
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.model_selection import cross_val_score
from sklearn.impute import SimpleImputer
from sklearn.impute import SimpleImputer
from sklearn.metrics import f1_score,make_scorer
from sklearn.model_selection import StratifiedKFold, cross_val_score
```

```python
# 1. Train-test split first
X_full = x.copy()  # `x` is your full feature set
y_full = y.copy()  # target variable

X_train, X_test, y_train, y_test = train_test_split(X_full, y_full, stratify=y, test_size=0.2, random_state=42)


# 2. Identify column types from **training set only**
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns
categorical_features = X_train.select_dtypes(include=['object']).columns

# 3. Separate into numeric and categorical
X_train_num = X_train[numeric_features]
X_test_num = X_test[numeric_features]

X_train_cat = X_train[categorical_features]
X_test_cat = X_test[categorical_features]

num_imputer = SimpleImputer(strategy='mean')
cat_imputer = SimpleImputer(strategy='most_frequent')

X_train[numeric_features] = num_imputer.fit_transform(X_train[numeric_features])
X_test[numeric_features] = num_imputer.transform(X_test[numeric_features])

X_train[categorical_features] = cat_imputer.fit_transform(X_train[categorical_features])
X_test[categorical_features] = cat_imputer.transform(X_test[categorical_features])


# 4. Scale numeric features using only training data
scaler = StandardScaler()
X_train_scaled = pd.DataFrame(
    scaler.fit_transform(X_train_num),
    columns=numeric_features,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    scaler.transform(X_test_num),
    columns=numeric_features,
    index=X_test.index
)

# 5. One-hot encode categorical features using **get_dummies + align**
X_train_encoded = pd.get_dummies(X_train_cat, drop_first=True)
X_test_encoded = pd.get_dummies(X_test_cat, drop_first=True)

# Align the test set to the train set’s columns
X_train_encoded, X_test_encoded = X_train_encoded.align(X_test_encoded, join='left', axis=1, fill_value=0)

# 6. Concatenate numeric and categorical
X_train_final = pd.concat([X_train_scaled, X_train_encoded], axis=1)
X_test_final = pd.concat([X_test_scaled, X_test_encoded], axis=1)

from imblearn.over_sampling import SMOTE

smote = SMOTE(
    random_state=42,
    k_neighbors=2   # safe for very small classes
)

X_train_resampled, y_train_resampled = smote.fit_resample(
    X_train_final, y_train
)
```

```python
# Define models
models = []
models.append(('LR', LogisticRegression(solver='lbfgs', multi_class='auto', max_iter=2000))) #, class_weight='balanced'
models.append(('KNN', KNeighborsClassifier()))
models.append(('NB', GaussianNB()))
models.append(('SVC', SVC(gamma='scale'))) #, class_weight='balanced'
models.append(('RFC', RandomForestClassifier(n_estimators=100))) #, class_weight='balanced'
models.append(('DTC', DecisionTreeClassifier())) #class_weight='balanced'
models.append(('GBC', GradientBoostingClassifier()))

# Evaluate models
results = []
names = []
for name, model in models:
    try:
        model.fit(X_train_resampled, y_train_resampled)
        y_pred_tr = model.predict(X_train_resampled)
        y_pred = model.predict( X_test_final)
        train_acc = accuracy_score(y_train_resampled, y_pred_tr)
        accuracy = accuracy_score(y_test, y_pred)
        print(f"Model: {name}, Accuracy: {train_acc}")
        print(f"Model: {name}, test Accuracy: {accuracy}")
        print(classification_report(y_test, y_pred))
        results.append(accuracy)
        names.append(name)
    except Exception as e:  # Handle potential errors during model training/prediction
        print(f"Error with model {name}: {e}")
        results.append(0)
        names.append(name)
```

```python
from sklearn.ensemble import RandomForestClassifier

# rfc = RandomForestClassifier(
#     n_estimators=500,
#     max_depth=None,
#     min_samples_leaf=3,
#     #class_weight='balanced',
#     random_state=SEED
# )

# #svc = SVC(gamma='scale',random_state=SEED)
# svm = SVC(probability=True, class_weight='balanced', gamma='scale')

# gbc =  GradientBoostingClassifier()

# #lr = LogisticRegression( random_state=SEED) #solver='lbfgs', multi_class='auto', max_iter=2000,


rfc_tuned = RandomForestClassifier(
    n_estimators=400,
    max_depth=12,
    min_samples_leaf=4,
    min_samples_split=6,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

svc_tuned = SVC(
    kernel='rbf',
    C=2,
    gamma='scale',
    probability=True,
    class_weight='balanced'
)

gbc_tuned = GradientBoostingClassifier(
    n_estimators=250,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.85,
    random_state=42
)
```

```python
classifiers = [('RandomForestClassifier',rfc_tuned),
               ('svc',svc_tuned),
               ('GradientBoostingClassifier',gbc_tuned)]

vc = VotingClassifier(estimators = classifiers,  voting='soft',
    weights=[2, 2, 3]  # RF strongest)
                     )
```

```python
# # Identify column types
# numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns
# categorical_features = X_train.select_dtypes(include=['object']).columns

# # Pipelines for preprocessing
# numeric_transformer = Pipeline(steps=[
#     ('scaler', StandardScaler())  # optional for RandomForest
# ])

# categorical_transformer = Pipeline(steps=[
#     ('onehot', OneHotEncoder(handle_unknown='ignore'))
# ])

#  # Combine preprocessing
# preprocessor = ColumnTransformer(transformers=[
#     ('num', numeric_transformer, numeric_features),
#     ('cat', categorical_transformer, categorical_features)
# ])

#             # Full pipeline
# pipeline = Pipeline(steps=[
#     ('preprocessor', preprocessor),
#     ('model', RandomForestClassifier(
#     n_estimators=100,
#     max_depth=None,
#     min_samples_leaf=2,
#     #class_weight='balanced',
#     random_state=42
# )
# )
# ])     
```

```python
# Identify column types
numeric_features = X_train.select_dtypes(include=['int64', 'float64']).columns
categorical_features = X_train.select_dtypes(include=['object']).columns

# Pipelines for preprocessing
numeric_transformer = Pipeline(steps=[
    ('scaler', StandardScaler())  # optional for RandomForest
])

categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

 # Combine preprocessing
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

            # Full pipeline
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', vc)
])
```

```python
pipeline.fit(X_train, y_train)
```

```python
score = pipeline.score(X_test, y_test)
score
```

```python
tr_score = pipeline.score(X_train, y_train)
tr_score
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
f1_weighted = make_scorer(f1_score, average='weighted')
```

```python
cv_scores = cross_val_score(
    pipeline,
    X_full,
    y_full,
    cv=cv,
    scoring=f1_weighted,
    n_jobs=-1
)
```

```python
print("Cross-Validated F1 Scores:", cv_scores)
print("Mean F1 Score:", np.mean(cv_scores))
print("Standard Deviation:", np.std(cv_scores))
```
